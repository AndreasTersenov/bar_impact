```python
w = mrs_uwttrans(inmap, verbose=False) # compute the wavelet transform of the map
```

```python
w.shape
for filter_ in range(w.shape[0]):
    fig = plt.figure(filter_)
    hp.mollview(w[filter_, :], title=f'Wavelet {filter_}', cbar=False, notext=True, coord='C')
    plt.title(f'Wavelet {filter_}', color='white')
    plt.savefig(f"./plots_dir/Wavelet_{filter_}.pdf", transparent=True, bbox_inches='tight')
    plt.show()
```
