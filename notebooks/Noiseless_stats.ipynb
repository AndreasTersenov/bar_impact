```python
import os
import numpy as np
import glob
import matplotlib.pyplot as plt
```

```python
rebin_factor = 10
# rebin by averaging every rebin_factor elements
def rebin(a, factor):
    sh = a.shape[0]//factor, factor
    return a[:sh[0]*factor].reshape(sh).mean(-1)
```

### Auto fiducial

```python
auto_fiducials_1 = np.load("/home/tersenov/CosmoGridV1/stage3_forecast/fiducial/cosmo_fiducial/all_cls_fiducial_nobaryons_bin1.npy")
auto_fiducials_2 = np.load("/home/tersenov/CosmoGridV1/stage3_forecast/fiducial/cosmo_fiducial/all_cls_fiducial_nobaryons_bin2.npy")
auto_fiducials_3 = np.load("/home/tersenov/CosmoGridV1/stage3_forecast/fiducial/cosmo_fiducial/all_cls_fiducial_nobaryons_bin3.npy")
auto_fiducials_4 = np.load("/home/tersenov/CosmoGridV1/stage3_forecast/fiducial/cosmo_fiducial/all_cls_fiducial_nobaryons_bin4.npy")
auto_fiducials_1.shape
```

```python
rebinned_auto_fiducials_1 = np.array([rebin(auto_fiducials_1[i], rebin_factor) for i in range(auto_fiducials_1.shape[0])])
rebinned_auto_fiducials_2 = np.array([rebin(auto_fiducials_2[i], rebin_factor) for i in range(auto_fiducials_2.shape[0])])
rebinned_auto_fiducials_3 = np.array([rebin(auto_fiducials_3[i], rebin_factor) for i in range(auto_fiducials_3.shape[0])])
rebinned_auto_fiducials_4 = np.array([rebin(auto_fiducials_4[i], rebin_factor) for i in range(auto_fiducials_4.shape[0])])
rebinned_auto_fiducials_1.shape
```

```python
fid_11 = np.mean(rebinned_auto_fiducials_1, axis=0)
fid_22 = np.mean(rebinned_auto_fiducials_2, axis=0)
fid_33 = np.mean(rebinned_auto_fiducials_3, axis=0)
fid_44 = np.mean(rebinned_auto_fiducials_4, axis=0)
```

```python
plt.loglog(fid_11, label='1,1')
plt.loglog(fid_22, label='2,2')
plt.loglog(fid_33, label='3,3')
plt.loglog(fid_44, label='4,4')
plt.legend()
plt.ylim(1e-11, 1e-7)
plt.show()
```
