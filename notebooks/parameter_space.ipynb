```python
import os
import matplotlib.pyplot as plt
import numpy as np
```

```python
params = np.load('/home/tersenov/CosmoGridV1/stage3_forecast/grid/cosmo_params.npy', allow_pickle=True)
```

```python
params.shape
```

```python
# identify the unique rows in params
unique_params = np.unique(params, axis=0)
unique_params.shape
```

```python
# save the unique rows to a new file
np.save('unique_cosmo_params.npy', unique_params)
```

```python
# plot the first two columns of unique_params
plt.figure(figsize=(8, 6))
plt.scatter(unique_params[:, 0], unique_params[:, 1], alpha=0.5)
plt.xlabel('Parameter 1')
plt.ylabel('Parameter 2')
plt.title('Scatter plot of first two parameters')
plt.grid(True)
plt.show()
```

```python
# plot all the parameters one by one
num_params = unique_params.shape[1]
fig, axes = plt.subplots(1, num_params, figsize=(5 * num_params,
    4))
for i in range(num_params):
    axes[i].hist(unique_params[:, i], bins=30, alpha=0.7)
    axes[i].set_xlabel(f'Parameter {i + 1}')
    axes[i].set_ylabel('Frequency')
    axes[i].set_title(f'Histogram of Parameter {i + 1}')
plt.tight_layout()
plt.show()
```
